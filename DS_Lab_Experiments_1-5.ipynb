## Experiment 3 – Average sale price of houses with more than four bedrooms

**Scenario:** House data (read from a CSV into a NumPy array `house_data`) has columns: number of bedrooms, square footage, sale price.

**Aim:** Find the average sale price of houses having **more than 4 bedrooms**.

**Approach:** Create a boolean mask `bedrooms > 4`, use it to pick matching prices, then take `np.mean`.

> No house dataset was provided, so a sample `house_data.csv` is generated below and read back with `np.loadtxt`, exactly like importing a real CSV. If you have your own file, delete the generation cell and set `house_file` to your file name.

```python
# Generate a sample CSV (skip this cell if you have your own house CSV)
rng = np.random.default_rng(42)
n = 20
bedrooms = rng.integers(1, 8, size=n)
sqft = bedrooms * 600 + rng.integers(200, 900, size=n)
price = sqft * 150 + rng.integers(-20000, 20000, size=n)
sample = np.column_stack([bedrooms, sqft, price])
np.savetxt("house_data.csv", sample, delimiter=",", fmt="%d",
           header="bedrooms,square_footage,sale_price", comments="")
print("house_data.csv created")
```

```python
house_file = "house_data.csv"
house_data = np.loadtxt(house_file, delimiter=",", skiprows=1)   # skip header row
print("Shape:", house_data.shape)
print("First 5 rows (bedrooms, sqft, price):\n", house_data[:5])

bedrooms = house_data[:, 0]
sale_price = house_data[:, 2]

mask = bedrooms > 4                       # True for houses with more than 4 bedrooms
print("\nHouses with more than 4 bedrooms:", np.sum(mask))

if mask.any():
    print("Average sale price:", round(np.mean(sale_price[mask]), 2))
else:
    print("No house has more than 4 bedrooms.")
```

## Experiment 4 – Total yearly sales and % increase from Q1 to Q4

**Scenario:** Quarterly sales are stored in a NumPy array `sales_data` (one element per quarter).

**Aim:** Compute the total sales for the year and the percentage increase from Q1 to Q4.

**Formulas**
- Total = `np.sum(sales_data)`
- % increase = (Q4 − Q1) / Q1 × 100

```python
df4 = pd.read_csv(find_file("Sales (1).csv", "Sales__1_.csv", "Sales_(1).csv"))

# Month_sales looks like 'February 2023' -> convert to a date, keep only 2023, get quarter 1-4
dates = pd.to_datetime(df4["Month_sales"], format="%B %Y")
df4 = df4[dates.dt.year == 2023]
quarter = dates[dates.dt.year == 2023].dt.quarter.to_numpy()
total = df4["Total"].to_numpy()

# Add up the 'Total' of every sale in each quarter -> a 4-element NumPy array
sales_data = np.array([total[quarter == q].sum() for q in [1, 2, 3, 4]])
print("Quarterly sales (Q1..Q4):", np.round(sales_data, 2))

total_sales = np.sum(sales_data)
pct_increase = (sales_data[3] - sales_data[0]) / sales_data[0] * 100

print("Total sales for the year 2023:", round(total_sales, 2))
print(f"Percentage change from Q1 to Q4: {pct_increase:.2f}%")
print("(a negative value means sales decreased)")
```

## Experiment 5 – Average fuel efficiency and % improvement between two car models

**Scenario:** The fuel efficiency (miles per gallon) of different car models is stored in a NumPy array `fuel_efficiency`.

**Aim:** Calculate the average fuel efficiency and the percentage improvement between two models.

**Formulas**
- Average = `np.mean(fuel_efficiency)`
- % improvement = (new − old) / old × 100

The efficiency column in the file is `highway_mpg(Fuel Efficiency)`.

```python
fuel_df = pd.read_csv(find_file("Fuel_Efficiency_datasets.csv", "Fuel_Efficiency datasets.csv"))
print(fuel_df[["make", "model", "year", "fuel_type", "highway_mpg(Fuel Efficiency)"]].head())
print("Rows:", len(fuel_df))

fuel_efficiency = fuel_df["highway_mpg(Fuel Efficiency)"].to_numpy()
print("\nAverage fuel efficiency (all cars):", round(np.mean(fuel_efficiency), 2), "mpg")
```

```python
is_gas = (fuel_df["fuel_type"] == "gas").to_numpy()
print("Average fuel efficiency (gas cars only):", round(np.mean(fuel_efficiency[is_gas]), 2), "mpg")

# --- Percentage improvement between two car models ---
def pct_improvement(old, new):
    return (new - old) / old * 100

# Model 1: first car in the file; Model 2: the most efficient gas car
old_idx = 0
new_idx = np.where(is_gas)[0][np.argmax(fuel_efficiency[is_gas])]

old, new = fuel_efficiency[old_idx], fuel_efficiency[new_idx]
print(f"\nModel 1: {fuel_df.loc[old_idx, 'make']} {fuel_df.loc[old_idx, 'model']} -> {old} mpg")
print(f"Model 2: {fuel_df.loc[new_idx, 'make']} {fuel_df.loc[new_idx, 'model']} -> {new} mpg")
print(f"Percentage improvement: {pct_improvement(old, new):.2f}%")
```
